```python
n_cols = 3
n_plots = len(numerical_columns)
n_rows = int(np.ceil(n_plots / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(6 * n_cols, 4 * n_rows))
axes = np.array(axes).reshape(-1)

for i, num in enumerate(numerical_columns):
    sns.histplot(df[num], kde=True, ax=axes[i])
    axes[i].set_title(f"Histogram of {num}")

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# Robustly find the addiction-level column (ignores case, spaces, underscores).
normalized = {c.lower().replace(' ', '').replace('_', ''): c for c in df.columns}
target_key = 'addictionlevel'

if target_key not in normalized:
    raise KeyError(
        "Could not find 'Addiction Level' column. "
        f"Available columns: {list(df.columns)}"
    )

target_col = normalized[target_key]
Y = df[target_col]

# Use only numeric predictors for boxplots/correlation (exclude the target if numeric).
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
X = [col for col in numeric_cols if col != target_col]

# -------- Boxplots in one frame --------
n_cols = 3
n_plots = len(X)
n_rows = int(np.ceil(n_plots / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(10 * n_cols, 4 * n_rows))
axes = np.array(axes).reshape(-1)

for i, num in enumerate(X):
    sns.boxplot(x=Y, y=df[num], ax=axes[i])
    axes[i].set_title(f"Boxplot of {num} by {target_col}")
    axes[i].tick_params(axis='x', rotation=25)

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
plt.show()

```

```python
from sklearn.feature_selection import f_regression

X = df.drop(columns='Addiction_Level')
Y = df['Addiction_Level']

# f_regression only works on numeric columns - need to encode categoricals first
X_numeric = pd.get_dummies(X, drop_first=True)  # one-hot encode text/categorical columns

f_stats, p_values = f_regression(X_numeric, Y)

results = pd.DataFrame({
    'Feature': X_numeric.columns,
    'F_stat': f_stats,
    'p_value': p_values
}).sort_values('p_value')

print(results)
if results['p_value'].min() < 0.05:
    print("Some features are statistically significant predictors of Addiction_Level.")
else:
    print("No features are statistically significant predictors of Addiction_Level.")
```
